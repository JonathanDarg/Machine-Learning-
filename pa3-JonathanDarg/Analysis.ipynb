# Grid Search Analysis (Accuracy Scoring)

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load grid search results
results_acc = pd.read_csv("grid1_search_results_accuracy.csv")
results_bal = pd.read_csv("grid2_search_results_balanced_accuracy.csv")

# Replace NaN "param_class_weight" with string "None"
results_acc["param_class_weight"] = results_acc["param_class_weight"].fillna("None")
results_bal["param_class_weight"] = results_bal["param_class_weight"].fillna("None")

# Set plot style
sns.set(style="darkgrid")
```

###  1. How good do the results appear to be based on the scoring?

```python
sns.boxplot(data=results_acc, x="param_class_weight", y="mean_test_score")
```

When looking at the results based on the score is seems to have done alright as it is in the low 0.80's, still having room for improvement. And seeing how with no balancing the accuracy was better but this was becuase it was correctly predicting one class, but not so well for the other class.

### 2. How does the number of estimators affect overall scoring?


```python
sns.boxplot(data=results_acc, x="param_n_estimators", y="mean_test_score", hue="param_class_weight")
```

The number of estimators is the number of trees in the forest, but from 100 to 200 doesnt seem to have made much of a difference.


###  3. How does `max_features` relate to scoring?


```python
sns.boxplot(data=results_acc, x="param_max_features", y="mean_test_score", hue="param_class_weight")
```

`max_features` relate to scoring by balancing accuracy and generalization. and Making trees more random to reduce overfitting.

### 4. How does `max_features` affect the time it took to fit the model?


```python
sns.boxplot(data=results_acc, x="param_max_features", y="mean_fit_time", hue="param_class_weight")
```

`max_features` affect the time it took to fit the model because
of the mathematical operations it is doing seeing how log2 is
faster than sqrt in runtime. 

### 5. Does `oob_score` affect the goodness of the model?


```python
sns.boxplot(data=results_acc, x="param_oob_score", y="mean_test_score", hue="param_class_weight")
```

`oob_score` does not affect the goodness of the model as it not used during prediction or scoring in the GridSearch only providing an internal estimate of validation accuracy.

### 6. How does the hyperparameter you chose affect the results?


```python
sns.barplot(data=results_acc, x="param_max_depth", y="mean_test_score", hue="param_class_weight")
```

The hyperparameter that i chose was max_depth and when increasing the depth the score increased, alowing it to create a more complex model.
